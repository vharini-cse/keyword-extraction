```python
# Keyword Extraction using NLP

import re
from sklearn.feature_extraction.text import TfidfVectorizer

# Input document
text = """
Natural Language Processing is a branch of artificial intelligence.
It helps computers understand, process, and generate human language.
NLP is used in applications such as sentiment analysis, chatbots,
machine translation, speech recognition, and text classification.
"""

# Clean the text
clean_text = re.sub(r'[^a-zA-Z\s]', '', text.lower())

# TF-IDF
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform([clean_text])

# Get words and their scores
words = vectorizer.get_feature_names_out()
scores = tfidf_matrix.toarray()[0]

# Sort by TF-IDF score
keywords = sorted(
    zip(words, scores),
    key=lambda x: x[1],
    reverse=True
)

# Display top keywords
print("===== KEYWORD EXTRACTION =====\n")

print("Original Text:")
print(text)

print("\nTop Keywords:")

for word, score in keywords[:10]:
    print(f"{word} : {score:.4f}")
```
